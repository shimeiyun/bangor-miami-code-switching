# 01 Data audit

The first research task is to establish what the files contain before defining a model or statistic. This public notebook uses synthetic data but calls the same parsers as the licensed-corpus run.

## 1. Load the project modules

Keeping reusable code in `src/` prevents the notebook from becoming an untestable collection of copied functions.

```python
from pathlib import Path
import collections, sys
ROOT = Path.cwd().parent if Path.cwd().name == 'notebooks' else Path.cwd()
sys.path.insert(0, str(ROOT / 'src'))
from bangor_miami.chat import chat_inventory
from bangor_miami.tokens import iter_tsv_tokens
```

## 2. Inspect CHAT participant headers

The header inventory provides recording, speaker code, pseudonym, role, age and sex. It does not assume that CHAT codes equal questionnaire IDs.

```python
chat_rows = chat_inventory(ROOT / 'data/sample/chats')
{'speaker_recording_rows': len(chat_rows), 'columns': list(chat_rows[0]), 'rows': chat_rows}
```

## 3. Inspect token rows and language labels

TSV columns are accessed by name, so harmless changes in column order do not break the pipeline. Footer rows without speaker/filename identifiers are excluded.

```python
tokens = list(iter_tsv_tokens(ROOT / 'data/sample/tsvs'))
language_counts = collections.Counter(row['langid'] for row in tokens)
{'token_rows': len(tokens), 'language_counts': dict(language_counts), 'first_three_rows': tokens[:3]}
```
